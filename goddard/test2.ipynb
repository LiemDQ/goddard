```python
from functions import calc_isp_sl
from utils import to_si, g0
import cantera as ct
from pint import UnitRegistry
ureg = UnitRegistry()
Q_ = ureg.Quantity
```

```python

o_f_ratio = 6.0
temperature_h2 = Q_(20.270, 'K')
temperature_o2 = Q_(90.170, 'K')
pressure_chamber = Q_(3000, 'psi')

h2o2_filename = "./h2o2_react.yaml"

h2 = ct.Solution(h2o2_filename, 'liquid_hydrogen')
h2.TP = to_si(temperature_h2), to_si(pressure_chamber)

o2 = ct.Solution(h2o2_filename, 'liquid_oxygen')
o2.TP = to_si(temperature_o2), to_si(pressure_chamber)
h2()
o2()
```

```python
isp = calc_isp_sl(h2, o2, pressure_chamber, o_f_ratio, 35.0)[0]
print(f"ISP: {isp}s, {isp*g0} m/s")
```

```python
isp_frozen = calc_isp_sl(h2, o2, pressure_chamber, o_f_ratio, 35.0, True)[0]
print(f"ISP: {isp_frozen}s, {isp_frozen*g0} m/s")
```

```python
isp_equilibrium_cea = 4207.4 #based on expansion ratio of 35.0, hydrolox combustion at O/F of 6.0
isp_frozen_cea = 4081.3


def print_error(isp_cea, isp_goddard, name):
    percent_error = abs(isp_cea - isp_goddard)/isp_cea*100
    print(f"Results for {name} | CEA: {isp_cea:3.2f} m/s | calculated: {isp_goddard:3.2f} m/s | percent error: {percent_error:2.4f}%")

print_error(isp_equilibrium_cea, isp*g0, "hydrolox equilibrium")
print_error(isp_frozen_cea, isp_frozen*g0, "hydrolox frozen")
```
