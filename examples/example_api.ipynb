# API design philosophy

The API should be modular and have "granularity of functionality". Some ideas:

`Problem` objects should require data relevant to the problem at hand, can have idiosyncratic/very different constructors and would take care of 
sending the data to all the right places. 

Do we need a separate `solve` function call? What would it be good for? In Julia this is done so different choices of solver can be used, but we don't have 
different options for solvers here. Any options we would want to apply (e.g. frozen vs equilibrium nozzle) would modify the problem itself. 

Solver options involving the tolerances for the various loops? Could be a possibility, although would have to format that into a dict

## Internal objects
`Nozzle` for solving gas expansion problems

`Combustor` for performing combustion analysis



```python
# example of desired API

import goddard

# generate a rocket problem

prob = RocketProblem(
    fuel="Jet-A(L)",
    oxidizer="O2(L)",
    pressure=[5e6, 10e6, 15e6],
    mixture_ratio=OF_ratio([1.1,2.3,2.4]),
    super=[3,5,10,15,20],
    nozzle=frozen_nozzle(),
    combustor=infinite_area_combustor(),
)

result = solve(prob, solveroptions=None)

result.report() #prints out a summary of the problem

#see the Solution object in Julia's DifferentialEquations for a good example of an effective result API

result.isp #provides isp in a numpy array for convenient processing
result.supr #supersonic area ratio
```
